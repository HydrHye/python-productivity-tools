## **Using OpenAI's**
### To generate sentences for the words explored in dictionary

```python
 resp_ai = client.chat.completions.create(
        model="gpt-4o",
        messages=[
            {"role": "system", "content": "You are a helpful assistant that writes one sentence using a given word and incorporates characters meaningfully as named by the user."},
            {"role": "user", "content": f"Write a creative sentence using the word: {word} but using the characters Mavi if it is a girl, Caspy if it is a cat, and Safy, if it is a baby"},
        ]
    )
```

```python
sentence = resp_ai.choices[0].message.content.strip()
```

```python
print(sentence)
```

```python
 img_resp = client.images.generate(
        model="dall-e-3",
        prompt=f"Illustrate the scene as in the children's book. The scene is: {sentence}. FYI, Mavi is the mother, Caspy is a playful, fluffy, white Persian cat, and Safy is the baby boy.",
        n=1,
        size="1024x1024",
        quality="standard" #this part is different for newer image generator models.
    )
```

```python
image_url = img_resp.data[0].url
display(Image(url=image_url))
```
