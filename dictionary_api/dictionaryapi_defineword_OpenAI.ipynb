## **Using OpenAI's**
### To generate sentences for the words explored in dictionary

```python
 response_ai = client.chat.completions.create(
        model="gpt-4o",
        messages=[
            {"role": "system", "content": "You are a helpful assistant that writes one sentence using a given word and incorporates characters meaningfully as named by the user."},
            {"role": "user", "content": f"Write a creative sentence using the word: {word} but using the character's Mahvash if it is a girl, Casper if it is a cat, and Saifaan, if it is a baby"},
        ]
    )
```

```python
response_ai.choices[0].message.content.strip()
```
